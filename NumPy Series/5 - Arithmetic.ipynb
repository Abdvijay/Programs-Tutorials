```python
import numpy as np

arr = np.array([24,5,6,2])

print(arr + 100)
print(arr - 5)
print(arr * 10)
print(arr / 5)
print(arr ** 5)
```

```python
sqrt = np.sqrt(arr)
print(sqrt)
print(np.round(sqrt))
print(np.floor(sqrt))
print(np.ceil(sqrt))
```
