```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x = np.array([12, 10, 11, 14, 7, 6, 10, 14])
y = np.array([5000, 4000, 3000, 7000, 6000, 2000, 5000, 6800])

x1 = np.array([11, 12, 10, 15, 7, 6, 11, 15])
y1 = np.array([5000, 4060, 3060, 7400, 6060, 2060, 5006, 6840])

property = {
    "alpha" : 0.5, 
    "edgecolor" : "black"
}

plt.figure(figsize=(10,6))
plt.scatter(x,y, color="Green", **property, label = "Code IO")
plt.scatter(x1,y1, color="Blue", **property, label = "Logic IO")
plt.title("Subscribers count", fontfamily = "Algerian", fontsize = 10, color = "Red", pad= 15)
plt.legend()
plt.show()
```
