```python
import pandas as pd

df = pd.read_json("iris.json")
print(df)
```

```python
# Drop specific column

df = df.drop(columns=['sepalLength'])
print(df)
```

```python
# Data cleaning
dropDf = df.dropna(subset='sepalWidth')
print(dropDf)
```

```python
# Exchange data for NaN using fillna method

fillDf = df.fillna({'sepalWidth' : 3.0, 'sepalLength' : 4.7})
print(fillDf)
```

```python
# Replacing

fillDf['species'] = fillDf['species'].replace({
    "virginica" : "Vijay",
    "setosa" : "Dheeksha"
})
print(fillDf)
```

```python
# Standardizing

fillDf['species'] = fillDf['species'].str.lower()
print(fillDf)
```

```python
# Type conversion

import numpy as np
fillDf['sepalLength'] = fillDf['sepalLength'].astype(np.int64)
print(fillDf)
```

```python
# Remove duplications

print("Before removing duplicates")
df = pd.read_json("iris.json")
print(df)
print()

print("After removing duplicates")
df = df.drop_duplicates()
print(df)
```
