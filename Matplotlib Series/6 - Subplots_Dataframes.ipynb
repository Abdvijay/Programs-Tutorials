```python
import matplotlib.pyplot as plt
import numpy as np

# Figure - Whole canvas
# Axes - Subplot


figure, axes = plt.subplots(2,2)

# First Figure

years = np.array(
    [2021, 2022, 2023, 2024, 2025, 2026]
)
subs = np.array(
    [0, 10000, 60000, 100000, 200000, 300000]
)

axes[0,0].plot(years,subs, color = "Red")

# Second Figure

years = np.array(
    [2021, 2022, 2023, 2024]
)
subs = np.array(
    [0, 10000, 60000, 100000]
)

axes[0,1].plot(years, subs, color = "Blue")

# Third Figure

x = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
y = np.array([5000, 1000, 3000, 4500, 500, 7000])

axes[1,0].bar(x,y, color = "Green")

# Four Figure

x = np.array([12, 10, 11, 14, 7, 6, 10, 14])
y = np.array([5000, 4000, 3000, 7000, 6000, 2000, 5000, 6800])

x1 = np.array([11, 12, 10, 15, 7, 6, 11, 15])
y1 = np.array([5000, 4060, 3060, 7400, 6060, 2060, 5006, 6840])

axes[1,1].scatter(x,y, color = "skyblue")
axes[1,1].scatter(x1, y1, color = "Orange")

plt.tight_layout()

plt.show()
```
