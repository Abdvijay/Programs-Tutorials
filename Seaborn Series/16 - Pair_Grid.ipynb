```python
import seaborn as sns

tips_df = sns.load_dataset('tips')
tips_df.head()
```

```python
# Pairplot

sns.pairplot(tips_df, hue='sex')
```

```python
# Pair Grid
import matplotlib.pyplot as plt

iris_df = sns.load_dataset('iris')
# iris_df.head()
# sns.pairplot(iris_df)
iris_grid = sns.PairGrid(iris_df)
# iris_grid.map_offdiag(plt.scatter)
iris_grid.map_diag(plt.hist)
iris_grid.map_upper(sns.kdeplot)
iris_grid.map_lower(plt.scatter)
```

```python
# Customize

iris_df = sns.load_dataset('iris')
iris_grid = sns.PairGrid(iris_df, hue='species', x_vars=["petal_length", "sepal_length"],
                         y_vars=["petal_width", "sepal_width"])

iris_grid.map(sns.scatterplot)
iris_grid.add_legend()
```
