```python
import numpy as np
```

```python
# Zeros function

zeros = np.zeros((2))
print(zeros)
print()

zeros = np.zeros((2,2))
print(zeros)
print()

zeros = np.zeros((2,2,3))
print(zeros)
print()
```

```python
# Ones function

# Syntax :
#        arr_name = np.ones((no_of_array_shape if one dimensional just mentioned total values))

ones = np.ones((2)) # here one dimensional with 2 values
ones = np.array(ones,dtype = np.str_)
print(ones)
print()

ones = np.ones((2,2)) # here 2 x 2 matrix
print(ones)
print()

ones = np.ones((2,2,3)) # here we created 2 layers with 2 rows and 3 columns
print(ones)
print()
```

```python
# Full function

# Syntax :
#        arr_name = np.full((no_of_array_shape if one dimensional just mentioned total values), value_with_filled)

arr = np.full((2),2)
print(arr)
print()

arr = np.full((2,2),2)
print(arr)
print()

arr = np.full((2,2,3),100)
print(arr)
print()
```

```python
# Identity matrix

arr = np.eye(3)
print(arr)
print()

arr = np.eye(2)
print(arr)
print()

arr = np.eye(5)
print(arr)
print()

arr = np.eye(1)
print(arr)
print()
```

```python
# Empty function

arr = np.empty(2)
print(arr)
print()

arr = np.empty((2,3))
print(arr)
print()
```

```python
# A range function

arr = np.arange(0,100,5)
print(arr)
print()

arr = np.arange(0,100,8)
print(arr)
print()
```

```python
# Linspace function

# Note : Here end value should also included

arr = np.linspace(0,100,6)
print(arr)
print()

arr = np.linspace(0,100,13)
print(arr)
print()
```

```python
# Aggragate functions

arr = np.arange(0,20,3)
print(arr)
print()

print(f"Minimum value : {arr.min()}")
print(f"Maximum value : {arr.max()}")
# print(f"Average value : {arr.avg()}") # Here avg function wont work because that is python aggragate function thats why
print(f"Average value : {arr.mean()}")
print(f"Median value : {np.median(arr)}")
print(f"Sum value : {arr.sum()}")
print(f"Stadard deviation value : {arr.std()}")
print(f"Variance value : {arr.var()}")
print(f"Minimum value position : {arr.argmin()}")
print(f"Maximum value position : {arr.argmax()}")
```

```python
# Matrix

arr = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr)
print()

print(np.sum(arr,axis = 1)) # Row wise sum
print()

print(np.sum(arr,axis = 0)) # Column wise sum
print()
```
