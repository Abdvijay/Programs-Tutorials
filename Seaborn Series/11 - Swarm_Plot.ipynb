```python
import seaborn as sns

tips_df = sns.load_dataset('tips')
tips_df.head()
```

```python
# Swarm plot
import matplotlib.pyplot as plt

sns.swarmplot(x='day', y='total_bill', hue='sex',  data=tips_df)
plt.legend(loc=0)
plt.savefig("Swarm-Plot.jpg")
```

```python
# Swarm plot with violin plot
import matplotlib.pyplot as plt

sns.violinplot(x='day', y='total_bill', hue='sex',  data=tips_df, split=True)
sns.swarmplot(x='day', y='total_bill', hue='sex',  data=tips_df, color='white')
plt.legend(loc=0)
```
