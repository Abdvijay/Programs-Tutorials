```python
import pandas as pd

age = [12,45,32,45]
series = pd.Series(age)
print(series)

# Series creates index for each value automatically
```

```python
age = ['vijay',22,True]
series = pd.Series(age)
print(series)

# So you use any type of data
```

```python
# You can set manually index using index property

age = [23,34,23,56]
series = pd.Series(age, index=['vijay','rajan','swathi','daddy'])
print(series)
```

```python
# loc -> It returns values using index
# iloc -> It always returns 2nd value

print(series.loc["vijay"])
print(series.iloc[1])
```

```python
# Dictionary used in series

age = {
    "vijay" : 23,
    "swathi" : 24,
    "dheeksha" : 2
}

series = pd.Series(age)
print(series)
```
