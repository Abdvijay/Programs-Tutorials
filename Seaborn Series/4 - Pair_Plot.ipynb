```python
import seaborn as sns

# Definition - Took one column and compare with another column known as pair plot.
cars = sns.load_dataset('tips')
cars.head()
```

```python
import matplotlib.pyplot as plt
sns.pairplot(cars)
plt.savefig("Pair-Plot.jpg")
```

```python
# Note : It used only 3 columns not all columns
sns.pairplot(cars, hue='sex')
```

```python
# Using palette color

sns.pairplot(cars, hue='sex', palette = 'Blues')
plt.savefig("Pair-Plot.jpg")
```
