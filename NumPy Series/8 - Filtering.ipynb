```python
import numpy as np

marks = np.array([59,23,90,34,67])
print(marks)
print()

print(marks[marks<=50])
print()

print(marks[(marks <= 50) & (marks % 2 == 0)])
print()

print(marks[marks % 2 == 0])
print()

print(marks % 2 == 0)
print()
```

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

print(matrix)
print()

even = matrix[matrix % 2 == 0]
print(even)
print()

# Note : Above filter even number from matrix but it always returns flatten matrix
```

```python
print(np.where(matrix % 2 == 0, matrix, -1))
print()

# Note :  NumPy where function gives same shape with filtering
# Syntax : np.where(filtering condition, performed matrix, filling value)
```
