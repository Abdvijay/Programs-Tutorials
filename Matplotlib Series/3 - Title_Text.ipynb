```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array(
    [2021, 2022, 2023, 2024, 2025, 2026]
)
subs = np.array(
    [0, 10000, 60000, 100000, 200000, 300000]
)

subs1 = np.array([0, 1000, 2000, 4000, 6000,10000])

plotPros = dict (
        marker = "v", 
        markersize=10, 
        markerfacecolor = "darkgrey",
        markeredgecolor = "red",
        markeredgewidth = 3,
        linestyle = "solid",
        linewidth = 5,
)

plt.plot(years, subs, color="black", **plotPros)

plt.plot(years, subs1, color="blue", **plotPros)

plt.title("Subscriber Counts",
          fontsize = 10,
          fontweight = "bold",
          color = "red",
          fontfamily = "Algerian"
          
)

plt.xlabel("Year wise", fontsize = 10, fontfamily = 'arial', color = "green")
plt.ylabel("Subscribers count", fontsize = 10, fontfamily = 'arial', color = "green")

plt.subplots_adjust(left=0.145, bottom=0.15)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array(
    [2021, 2022, 2023, 2024]
)
subs = np.array(
    [0, 10000, 60000, 100000]
)

subs1 = np.array([0, 1000, 2000, 4000])

plotPros = dict (
        marker = "v", 
        markersize=10, 
        markerfacecolor = "darkgrey",
        markeredgecolor = "red",
        markeredgewidth = 3,
        linestyle = "solid",
        linewidth = 5,
)

plt.plot(years, subs, color="black", **plotPros)

plt.plot(years, subs1, color="blue", **plotPros)

plt.title("Subscriber Counts",
          fontsize = 10,
          fontweight = "bold",
          color = "red",
          fontfamily = "Algerian",
          pad = 30
          
)

plt.xlabel("Year wise", fontsize = 10, fontfamily = 'arial', color = "green")
plt.ylabel("Subscribers count", fontsize = 10, fontfamily = 'arial', color = "green")

plt.subplots_adjust(left=0.145, bottom=0.15)

plt.xticks(years)
plt.tick_params(axis="x", labelsize = 10, labelcolor = "grey") # axis = "both"
plt.tick_params(axis="y", labelsize = 10, labelcolor = "grey")

plt.grid(axis="both", color = "brown", linestyle="dotted", linewidth=1)
# plt.grid(axis="x")
# plt.grid(axis="y")

plt.show()
```
