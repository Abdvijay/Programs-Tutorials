```python
import numpy as np

rng = np.random.default_rng()
print(rng.integers(1,31))
print()

print(rng.integers(low=1,high=31,size=5))
print()

print(rng.integers(low=0,high=100,size=(3,3)))
print()

print(np.random.default_rng().integers(1,34,size=(5)))
print()
```

```python
# If you wnat you to repeat the same random numbers then set seed while initiating with random number

rng = np.random.default_rng(seed=23)
print(rng.integers(1,32,size=(3,3)))
print()
```

```python
# Floating point numbers generators

print(np.random.uniform(low=-1,high=20))
print()
```

```python
# Shuffle

rng = np.random.default_rng()

names = np.array(["vijay","dheeksha","swathi","diya","shenba"])
rng.shuffle(names)
print(names)
print()

# Choice - Pick only one

print(rng.choice(names))
```
