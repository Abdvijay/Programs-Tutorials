```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
y = np.array([5000, 1000, 3000, 4500, 500, 7000])

# plt.plot(x,y) # Line Chart
plt.bar(x,y) # Bar Chart
plt.show()
```

```python
plt.bar(x,y, color = "skyblue") # Vertical bar
# plt.barh(x, y, color = "skyblue") # Horizontal bar
plt.title("Month wise growth", color = "red", fontfamily = "Algerian", fontsize = 10, pad=20)
plt.xlabel("Month", color = "darkgreen")
plt.ylabel("Subscribers Count", color = "darkgreen")
plt.show()
```

```python
x = np.array(["Long form", "Shorts", "Live", "Posts"])
y = np.array([4060, 1200, 500, 200])

plt.pie(y, 
        labels=x, 
        autopct="%.2f%%",
        colors = ["red", "blue", "green", "yellow"],
        shadow=True,
        explode=[0, 0.1, 0, 0.2],
        startangle=90)

plt.title("Subscribers source", 
          color = "Green", 
          fontsize = 10, 
          fontfamily = "Algerian", 
          pad = 30)

plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# print(os.getcwd())

data = pd.read_csv("MONTH EXPENSE TRACKER - JUNE.csv", index_col="DATE")

# data.columns

data = data.fillna({
    'DATE': 0,
    'WATER': 0,
    'OIL': 0,
    'RICE': 0,
    'BANANA': 0,
    'SNACKS': 0,
    'BISCUIT': 0,
    'FOOD': 0,
    'EGG': 0,
    'ONION': 0,
    'TOMATO': 0,
    'POTATO': 0,
    'CHILLY': 0,
    'MAAVU': 0,
    'COCONUT': 0,
    'MILK': 0,
    'BUS FARE': 0,
    'CURD': 0,
    'DMART THINGS': 0,
    'NET BILL': 0,
    'EB BILL': 0,
    'RENT': 0,
    'SUGAR': 0,
    'MOVIE': 0,
    'RECHARGE': 0,
    'KITCHEN NEEDED THINGS': 0,
    'EXTRA THINGS': 0,
    'TRAIN TICKETS': 0,
    'DAY-WISE-TOTAL': 0
})

# data

x = data['DAY-WISE-TOTAL'].to_list()

y = [i.split("/")[0] for i in data.index.to_list()]

x = x[0:2] + x[3:9] + x[10 : -1]

y = y[0:2] + y[3:9] + y[10 : len(y)-1]

plt.figure(figsize=(15, 6))

plt.yticks(range(0, 1000, 100))

plt.bar(y, x, width=0.5)

plt.title("June Month Expenses", pad=30, fontsize = 20, fontfamily = "Algerian", color = "Red")
plt.xlabel("Day wise", color = "Green", fontsize = 20)
plt.ylabel("Rupees", color = "Green", fontsize = 20)

plt.xticks(rotation=45)

# plt.subplots_adjust(bottom=0.2)
plt.tight_layout()
```
