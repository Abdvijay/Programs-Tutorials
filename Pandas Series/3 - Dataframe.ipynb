```python
import pandas as pd

students = {
    "Names" : ["Vijay", "Swathi", "Dheeksha"],
    "Age" : [25,26,2],
    "District" : ["Chennai", "Nellai", "Nagercoil"]
}

df = pd.DataFrame(students)
print(df)
```

```python
df = pd.DataFrame(students, index=["Person 1", "Person 2", "Person 3"])
print(df)
print("")

print(df.loc["Person 1"])
print("")

print(df.iloc[1])
```

```python
# New field inserted

df["Gender"] = ["Boy", "Girl", "Girl"]
print(df)
```

```python
# New row inserted

person4 = pd.DataFrame({
    "Names" : "Devi",
    "Age" : 27,
    "District" : "Nagercoil",
    "Gender" : "Girl"
}, index=["Person 4"])

df = pd.concat([df, person4])
print(df)
```

```python
# Multiple data added

persons = pd.DataFrame(
    [{"Names" : "Deva", "Age" : 25, "District" : "Tenkasi", "Gender" : "Boy"},
     {"Names" : "Deva", "Age" : 25, "District" : "Tenkasi", "Gender" : "Boy"}],
    index=["Person 5", "Person 6"]
)

df = pd.concat([df, persons])
print(df)
```
