```python
import seaborn as sns

print(sns.get_dataset_names())

crashes_df = sns.load_dataset('car_crashes')
crashes_df.head()
```

```python
crashes_corr = crashes_df.select_dtypes(include=['number']).corr()
crashes_corr.head()
```

```python
# Heatmap

sns.heatmap(crashes_corr, cmap='Purples', annot=True)
```

```python
# Pivoting

flight_df = sns.load_dataset('flights')
flights = flight_df.pivot_table(index='year', columns='month', values='passengers')
sns.heatmap(flights, cmap='Blues')
```
