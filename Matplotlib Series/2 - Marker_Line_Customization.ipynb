```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array(
    [2021, 2022, 2023, 2024, 2025, 2026]
)
subs = np.array(
    [0, 10000, 60000, 100000, 200000, 300000]
)

plt.plot(years, subs, marker = ".")
plt.show()
```

```python
plt.plot(years, subs, 
         marker = "v", 
         markersize=10, 
         markerfacecolor = "darkgrey",
         markeredgecolor = "red",
         markeredgewidth = 3,
         linestyle = "solid",
         linewidth = 5,
         color = "black"
)

# linestyle = ["dashed" , "dotted", "dashdot", "solid", "None"]
```

```python
years = np.array(
    [2021, 2022, 2023, 2024, 2025, 2026]
)
subs = np.array(
    [0, 10000, 60000, 100000, 200000, 300000]
)

subs1 = np.array([0, 1000, 2000, 4000, 6000,10000])

plotPros = dict (
        marker = "v", 
        markersize=10, 
        markerfacecolor = "darkgrey",
        markeredgecolor = "red",
        markeredgewidth = 3,
        linestyle = "solid",
        linewidth = 5,
)

plt.plot(years, subs, color = "black", **plotPros)

plt.plot(years, subs1, color = "blue", **plotPros)
```
