```python
import pandas as pd

df = pd.read_json('iris.json')
print(df)
```

```python
# Average

print("Average")
print(df.mean(numeric_only=True))
print()

print("Maximum")
print(df.max(numeric_only=True))
print()

print("Minimum")
print(df.min(numeric_only=True))
print()
```

```python
print("Sum")
print(df.sum(numeric_only=True))
print()

print("Count")
print(df.count(numeric_only=True))
print()

print("Median")
print(df.median(numeric_only=True))
print()

print("Mode")
print(df.mode(numeric_only=True))
```

```python
print("Standard deviation")
print(df.std(numeric_only=True))
print()

print("Variance")
print(df.var(numeric_only=True))
print()
```

```python
# Specific column

print(f"Average of Petal Length -> {df["petalLength"].mean()}")
print(f"Maximum of Petal Length -> {df["petalLength"].max()}")
print(f"Maximum of Sepal Length -> {df["sepalLength"].sum()}")
print(f"Maximum of Sepal Width  -> {df["sepalWidth"].min()}")
```

```python
# Group by concept

group = df.groupby("species")

print("Mean")
print(group.mean())
print()

print("Sum")
print(group.sum())
print()

print("Average of Sepal length species wise")
print(group["sepalLength"].mean())
```

```python
# Sorting

# Ascending
print(f"Ascending order printing by sepalLength\n")
print(df.sort_values("sepalLength"))
print()

# Ascending
print(f"Descending order printing by petalLength\n")
print(df.sort_values("petalLength", ascending=False))
print()
```
