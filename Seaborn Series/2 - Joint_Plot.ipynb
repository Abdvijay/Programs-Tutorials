```python
import seaborn as sns

cars = sns.load_dataset('car_crashes')
cars.head()
```

```python
# Joint plot - Using Regression
sns.jointplot(x='alcohol', y='speeding', data= cars, kind='reg')
```

```python
# Joint Plot Using KDE - Kernal Distribution Estimation
sns.jointplot(x='alcohol', y='speeding', data= cars, kind='kde')
```

```python
# Joint Plot Using hex - Hexagon
sns.jointplot(x='alcohol', y='speeding', data= cars, kind='hex')
```
