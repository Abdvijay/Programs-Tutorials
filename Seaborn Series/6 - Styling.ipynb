```python
import seaborn as sns

car_crash_df = sns.load_dataset('car_crashes')
car_crash_df.head()
```

```python
# Styling

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
```

```python
# 1. White

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('white')
```

```python
# 2. whitegrid

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('whitegrid')
```

```python
# 3. dark

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('dark')
```

```python
# 4. darkgrid

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('darkgrid')
```

```python
# 5. ticks

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('ticks')
```

```python
# 6. size

import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_style('white')
```

```python
# 7. set context - paper

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_context('paper')
```

```python
# 8. set context - talk

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_context('talk')
```

```python
# 9. set context - poster

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_context('poster')
```

```python
# 10. font scale

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_context('paper', font_scale=1.5)
```

```python
# 11. despine()

sns.jointplot(x ='alcohol', y='speeding', data=car_crash_df)
sns.set_context('paper', font_scale=1.5)
sns.despine(left=True, bottom=True)
```
