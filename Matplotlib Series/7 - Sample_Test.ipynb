```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("iris.csv")

sepalLength = np.array(data["sepalLength"])
sepalWidth = np.array(data["sepalWidth"])
petalLength = np.array(data["petalLength"])
petalWidth = np.array(data["petalWidth"])

plt.figure(figsize=(10,6))
plt.scatter(sepalLength, sepalWidth, color = "red")
plt.scatter(petalLength, petalWidth, color = "blue")
plt.tight_layout()
plt.title("Iris Data Set", fontsize = 15, color = "Green", fontfamily = "Algerian", pad = 15)

plt.show()
```
